## load model

```python
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

load_dotenv()

model = init_chat_model(
    "openai:gpt-5.6-luna", 
    use_responses_api=True,
    reasoning_effort="low",
)
```

```python
# test if model loads
response = model.invoke("Reply with exactly: OK")
print(response.content)
```

Deep agent

```python
from deepagents import create_deep_agent

agent = create_deep_agent(
    model=model,
    system_prompt=(
        "You are a helpful research paper review assistant."
    )
)
```

```python
result = agent.invoke({
    "messages": [{
        "role": "user",
        "content": (
            "What are the latest papers about self-improving agents?"
        )
    }]
})

print(result)
```

```python
print(result["messages"][-1].content)
```

```python
import os
from pathlib import Path
from dotenv import load_dotenv

print("Working directory:", Path.cwd())
load_dotenv()  # Loads .env from the repository
```

```python
from paper_research import create_research_agent

agent = create_research_agent(
    model="openai:gpt-5.6-luna",
    workspace_dir="research_data/notebook",
)
```

```python
result = agent.invoke(
    {
        "messages": [
            {
                "role": "user",
                "content": (
                    "Read arXiv paper 1706.03762v7. "
                    "Summarize its main method and provide PDF page citations."
                ),
            }
        ]
    },
    config={"recursion_limit": 50},
)

print(result["messages"][-1].content)
```

```python
from pprint import pprint
pprint(result["messages"][-1].content)
```

```python
pprint(result["messages"][-1].content[1]['text'])
```

```python
for index, message in enumerate(result["messages"]):
    print(f"\n--- {index}: {type(message).__name__} ---")

    tool_calls = getattr(message, "tool_calls", None)
    if tool_calls:
        print("Tool calls:", tool_calls)

    content = getattr(message, "content", "")
    if content:
        print(str(content)[:2000])
```
